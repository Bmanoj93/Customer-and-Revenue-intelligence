```python
%sql
-- 1. Total Revenue
SELECT SUM(Revenue) AS TotalRevenue
FROM fact_sales;
```

```python
%sql
-- 2. Total Profit
SELECT SUM(Profit) AS TotalProfit
FROM fact_sales;
```

```python
%sql
-- 3. Total Orders
SELECT COUNT(DISTINCT OrderNumber) AS TotalOrders
FROM fact_sales;
```

```python
%sql
-- 4. Total Customers
SELECT COUNT(DISTINCT CustomerKey) AS TotalCustomers
FROM fact_sales;
```

```python
%sql
-- 5. Revenue by Year
SELECT
    YEAR(OrderDate) AS Year,
    SUM(Revenue) AS Revenue
FROM fact_sales
GROUP BY YEAR(OrderDate)
ORDER BY Year;
```

```python
%sql
-- 6. Revenue by Category
SELECT
    p.CategoryName,
    SUM(f.Revenue) AS Revenue
FROM fact_sales f
JOIN dim_product p
    ON f.ProductKey = p.ProductKey
GROUP BY p.CategoryName
ORDER BY Revenue DESC;
```

```python
%sql
-- 7. Revenue by Territory
SELECT
    t.Region,
    SUM(f.Revenue) AS Revenue
FROM fact_sales f
JOIN dim_territory t
    ON f.TerritoryKey = t.SalesTerritoryKey
GROUP BY t.Region
ORDER BY Revenue DESC;
```

```python
%sql
-- 8. Top 10 Customers
SELECT
    CustomerKey,
    SUM(Revenue) AS Revenue
FROM fact_sales
GROUP BY CustomerKey
ORDER BY Revenue DESC
LIMIT 10;
```

```python
%sql
-- 9. Top 10 Products
SELECT
    ProductKey,
    SUM(Revenue) AS Revenue
FROM fact_sales
GROUP BY ProductKey
ORDER BY Revenue DESC
LIMIT 10;
```

```python
%sql
-- 10. Monthly Revenue
SELECT
    YEAR(OrderDate) AS Year,
    MONTH(OrderDate) AS Month,
    SUM(Revenue) AS Revenue
FROM fact_sales
GROUP BY YEAR(OrderDate), MONTH(OrderDate)
ORDER BY Year, Month;
```

```python
%sql
-- 11. Customers with more than 10,000 revenue
SELECT
    CustomerKey,
    SUM(Revenue) AS TotalRevenue
FROM fact_sales
GROUP BY CustomerKey
HAVING SUM(Revenue) > 10000
ORDER BY TotalRevenue DESC;
```

```python
%sql
-- 12. Return Quantity by Product
SELECT
    ProductKey,
    SUM(ReturnQuantity) AS TotalReturned
FROM fact_returns
GROUP BY ProductKey
ORDER BY TotalReturned DESC;
```

```python
%sql
-- 13. Churned Customers
SELECT
    COUNT(DISTINCT CustomerKey) AS ChurnedCustomers
FROM gold_customer_churn
WHERE ChurnStatus = 'Churned';
```

```python
%sql
-- 14. Churn Rate
SELECT
    ROUND(
        100.0 * SUM(CASE WHEN ChurnStatus = 'Churned' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS ChurnRate
FROM gold_customer_churn;
```

```python
%sql
-- 15. Complete customer revenue analysis
SELECT
    c.CustomerKey,
    c.FirstName,
    c.LastName,
    SUM(f.Revenue) AS TotalRevenue,
    SUM(f.Profit) AS TotalProfit,
    COUNT(DISTINCT f.OrderNumber) AS TotalOrders
FROM fact_sales f
JOIN dim_customer c
    ON f.CustomerKey = c.CustomerKey
GROUP BY
    c.CustomerKey,
    c.FirstName,
    c.LastName
ORDER BY TotalRevenue DESC;
```
